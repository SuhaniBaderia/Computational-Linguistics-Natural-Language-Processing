1. Create a small dataset containing positive, negative, and neutral sentences.
2. Preprocess the text by converting it to lowercase and removing punctuation and stop words and other required steps.
3. Count the positive and negative words in each sentence.
4. Assign sentiment labels using the following rule:
    (a) Positive score > threshold: Positive
    (b) Negative score < threshold: Negative
    (c) Otherwise: Neutral
5. Display the sentiment score and predicted sentiment for each sentence.
6. Evaluate the system using accuracy (precision, recall, f1score).


```python
import re
import nltk
import pandas as pd
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report

nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")
```

1. Create a small dataset containing positive, negative, and neutral sentences


```python
dataset = [
    ("I loved the wonderful movie and the brilliant acting.", "Positive"),
    ("The food was amazing and the service was excellent.", "Positive"),
    ("We enjoyed a beautiful and pleasant evening together.", "Positive"),
    ("This is a fantastic book with an outstanding story.", "Positive"),
    ("I hated the terrible movie and the awful acting.", "Negative"),
    ("The food was horrible and the service was poor.", "Negative"),
    ("We had a boring and disappointing evening together.", "Negative"),
    ("This is a dreadful book with an unpleasant story.", "Negative"),
    ("The book is on the table near the window.", "Neutral"),
    ("Students attend the lecture in the morning.", "Neutral"),
    ("The train arrives at the station at noon.", "Neutral"),
    ("The report contains numbers and charts about sales.", "Neutral"),
]

print("Sentence | Actual sentiment")
for sentence, label in dataset:
    print(sentence, "|", label)
```

2. Preprocess the text by converting it to lowercase and removing punctuation, stop words, and tokenizing


```python
stop_words = set(stopwords.words("english"))

def preprocess(text):
    text = text.lower()
    text = re.sub(r"[^a-z\s]", "", text)
    tokens = word_tokenize(text)
    tokens = [word for word in tokens if word not in stop_words]
    return tokens

print("Original sentence")
print("Preprocessed tokens")
print()
for sentence, label in dataset:
    print(sentence)
    print(preprocess(sentence))
    print()
```

3. Count the positive and negative words in each sentence

The sentiment score is the positive word count minus the negative word count. The threshold is 0.

4. Assign sentiment labels using the following rule:
(a) Sentiment score > threshold: Positive
(b) Sentiment score < threshold: Negative
(c) Otherwise: Neutral


```python
positive_words = {
    "love", "loved", "wonderful", "brilliant", "amazing", "excellent",
    "enjoyed", "beautiful", "pleasant", "fantastic", "outstanding",
    "good", "great", "happy", "best", "awesome", "delightful", "perfect"
}

negative_words = {
    "hate", "hated", "terrible", "awful", "horrible", "poor",
    "boring", "disappointing", "dreadful", "unpleasant",
    "bad", "worst", "sad", "ugly", "angry", "disgusting", "failure"
}

threshold = 0

def score_sentence(tokens):
    found_positive = [word for word in tokens if word in positive_words]
    found_negative = [word for word in tokens if word in negative_words]
    positive_score = len(found_positive)
    negative_score = len(found_negative)
    sentiment_score = positive_score - negative_score

    if sentiment_score > threshold:
        predicted = "Positive"
    elif sentiment_score < threshold:
        predicted = "Negative"
    else:
        predicted = "Neutral"

    return found_positive, found_negative, positive_score, negative_score, sentiment_score, predicted
```

5. Display the sentiment score and predicted sentiment for each sentence


```python
rows = []
for sentence, actual in dataset:
    tokens = preprocess(sentence)
    found_positive, found_negative, positive_score, negative_score, sentiment_score, predicted = score_sentence(tokens)
    rows.append({
        "Sentence": sentence,
        "Positive words": ", ".join(found_positive) if found_positive else "-",
        "Negative words": ", ".join(found_negative) if found_negative else "-",
        "Positive score": positive_score,
        "Negative score": negative_score,
        "Sentiment score": sentiment_score,
        "Predicted sentiment": predicted,
        "Actual sentiment": actual,
    })

results = pd.DataFrame(rows)
print(results.to_string(index=False))
```

6. Evaluate the system using accuracy, precision, recall, and F1-score


```python
y_true = results["Actual sentiment"].tolist()
y_pred = results["Predicted sentiment"].tolist()
labels = ["Positive", "Negative", "Neutral"]

accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred, labels=labels, average="weighted", zero_division=0)
recall = recall_score(y_true, y_pred, labels=labels, average="weighted", zero_division=0)
f1 = f1_score(y_true, y_pred, labels=labels, average="weighted", zero_division=0)

print("Accuracy  :", round(accuracy, 4))
print("Precision :", round(precision, 4))
print("Recall    :", round(recall, 4))
print("F1-score  :", round(f1, 4))
print()
print(classification_report(y_true, y_pred, labels=labels, zero_division=0))
```
